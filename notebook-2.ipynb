````python
# Merge the data for calls, minutes, messages, internet based on user_id and month

calls_agg = calls.groupby(['user_id' , 'month']).agg({'duration': 'sum', 
                                                       'id': 'count'
                                                       }).reset_index()
calls_agg.columns = ['user_id', 'month', 'total_minutes', 'total_calls']


messages_agg = messages.groupby(['user_id', 'month']).size().reset_index(name='total_messages')


internet_agg = internet.groupby(['user_id', 'month'])['mb_used'].sum().reset_index()
internet_agg.columns = ['user_id', 'month', 'total_mb_used']

merged_df = calls_agg.merge(messages_agg, on=['user_id', 'month'], how='outer')
merged_df = merged_df.merge(internet_agg, on=['user_id', 'month'], how='outer')
````

````python


final_df['rounded_minutes'] = np. ceil(final_df['total_minutes'])
final_df['rounded_messages'] = final_df['total_messages']
final_df['total_gb_used'] = np. ceil(final_df['total_mb_used'] / 1024)

print(final_df[['user_id', 'month', 'total_minutes', 'rounded_minutes', 'total_messages', 'rounded_messages', 'total_mb_used', 'total_gb_used']] .head())

final_df['extra_minutes'] = (final_df['rounded_minutes'] - final_df['minutes_included']).clip(lower=0)
final_df['extra_messages'] = (final_df['rounded_messages'] - final_df['messages_included']).clip (lower=0) 

final_df['gb_included'] = final_df['mb_per_month_included'] / 1024
final_df['extra_gb'] = (final_df['total_gb_used'] - final_df['gb_included']).clip (lower=0) 

final_df['overage_charge_minutes'] = final_df['extra_minutes'] * final_df['usd_per_minute']
final_df['overage_charge_messages'] = final_df['extra_messages'] * final_df['usd_per_message']
final_df['overage_charge_data'] = final_df['extra_gb'] * final_df['usd_per_gb']

final_df['total_overage'] = (
    final_df['overage_charge_minutes'] + 
    final_df['overage_charge_messages'] + 
    final_df['overage_charge_data']
)

def calculate_monthly_revenue(row):
    base_cost = row['usd_monthly_pay']
    overage = row['total_overage']
    return base_cost + overage

final_df['monthly_revenue'] = final_df.apply(calculate_monthly_revenue, axis=1)

print(final_df[['user_id', 'month', 'usd_monthly_pay', 'total_overage', 'monthly_revenue']].head(10))

revenue_by_plan = (
    final_df.groupby('plan_name')['monthly_revenue']
    .sum()
    .reset_index(name='total_revenue')
)

avg_revenue_by_plan = (
    final_df.groupby('plan_name')['monthly_revenue']
    .mean()
    .reset_index(name='avg_revenue_per_user')
)

problem_rows = final_df[final_df['monthly_revenue'] < final_df['usd_monthly_pay']]
print(f"rows where revenue < base cost: {len(problem_rows)}")

````
